## Definition

> A decision tree is a supervised learning algorithm used for classification and regression modeling. Decision trees look like flowcharts, starting at the root (like an upside-down tree) and branching off to demonstrate various outcomes. 

https://www.coursera.org/articles/decision-tree-machine-learning

> A decision tree is a supervised learning algorithm used for both classification and regression tasks. It has a hierarchical tree structure consisting of a root node, branches, internal nodes, and leaf nodes — working like a flowchart that helps in making step-by-step decisions.

https://www.geeksforgeeks.org/machine-learning/decision-tree-introduction-example/

- It's a supervised learning algorithm (it learns from labeled data)
- It works for both classification (categorizing) and regression (predicting numbers)
- It has a tree/flowchart structure — root → branches → leaves

An example of decision tree is as follows:

- Is it raining?
    - Yes → Wear a raincoat
    - No → Is it cold?
        - Yes → Wear a jacket
        - No → Wear a t-shirt

In machine learning, decision tree learn the answer to a question by breaking down the question into smaller simple questions. It picks questions that split the results most cleanly.

In classification problem, decision tree will find question that can split the labels (True/False) as clean as possible. We use Gini to calculate the impurity. This tell us how pure is the split. If the gini is 0, then all the labels in the node are the same. If we get 0.5, then it shows that there is a mix of positive and negative in the node.

In regression problem, it will find question that can reduce the variance/MSE (Mean Square Error) as much as possible.

## Tree Building Focus Using Gini as Threshold

In this section, we are going to find the best split point, which is the essence of decision tree. 


For classification problem, we use Gini impurity. Gini impurity calculates how pure is the labels in your dataset. If half your dataset contains true labels and the other half contains false label, then we can say the dataset is not pure. For a dataset which is not pure the Gini impurity is 0.5. If all labels are the same, then we can say the dataset is pure (Gini = 0).  

When we find the best threshold to split, instead of Gini, we calculate the weighted gini, to get the best split point. Why we use weighted gini? When finding a threshold to split, we need to split the dataset and calculate the gini of the split. We cannot use either the left or right split dataset. So the best is to use the weighted Gini to compute.  

### Gini Impurity Formula

For a node:

$$\text{Gini} = 1 - \sum_{k=1}^{K} p_k^2$$


Where:

- K = number of classes
- $p_k$ = proportion of class (k) in the node


#### Binary classification Example (0/1 case)

$$\text{Gini} = 1 - (p_0^2 + p_1^2)$$

Where:

- $p_0 = \frac{0}{N}$
- $p_1 = \frac{1}{N}$

**Intuition**

| Case           | Gini                          |
| -------------- | ----------------------------- |
| all same class | 0 (pure)                      |
| 50/50 split    | 0.5 (max impurity for binary) |

---

### Weighted Gini (for a split)

When you split a node into LEFT and RIGHT:

$$ \text{Weighted Gini} = \frac{N_L}{N} \cdot Gini(L) + \frac{N_R}{N} \cdot Gini(R)$$

Where:

- $N_L$ = number of samples in left node
- $N_R$ = number of samples in right node
- $N = N_L + N_R$


**Intuition**

Weighted Gini means:

> "how impure are the children, adjusted by their size?"

So:

- big bad child → matters a lot
- small bad child → matters less

#### Simple example

Left:

* 4 samples → Gini = 0

Right:

* 4 samples → Gini = 0.32

$$Weighted = \frac{4}{8}(0) + \frac{4}{8}(0.32) = 0.16$$


### Key insight

**Gini answers:**

> “How mixed is one group?”

**Weighted Gini answers:**

> “How good is this split overall?”



### One-line summary

```text
Gini = impurity of a node
Weighted Gini = impurity of a split
```

### Gini Split Search Procedure:
- Sort feature values
- For each adjacent pair of feature value:
    - compute midpoint threshold
- For each threshold:
    - split data
    - compute weighted Gini
- Choose best threshold

> Note: In Gini Split search, we use weighted Gini instead.

## Tree Building Focus Using Variance as Threshold

For regression problem, we use variance or mean square error instead of Gini. In production, we usually use square error which is more inline with our loss function in linear regression. We use variance for simplicity sake.


## Variance Formula

### Variance Formula

For a node:

$$\text{Variance} = \frac{\sum (y_i - \bar{y})^2}{n}$$


Where:

- $y_i$ = target values
- $\bar{y}$ = mean of target values


**Intuition**

| Case                    | Variance      |
| ----------------------- | ------------- |
| target tightly grouped  | low variance  |
| target widely scattered | high variance |

**A regression tree chooses splits that reduce variance after splitting**

---

### Weighted Variance (for a split)

When you split a node into LEFT and RIGHT:

$$ \text{Weighted Variance} = \frac{N_L}{N} \cdot Var(L) + \frac{N_R}{N} \cdot Var(R)$$

Where:

- $N_L$ = number of samples in left node
- $N_R$ = number of samples in right node
- $N = N_L + N_R$


**Intuition**

Weighted Variance means:

> "How much uncertainty remains after the split, considering how many samples fall into each side?"

### Variance split procedure:
- Sort feature values
- For each adjacent pair:
    - compute midpoint threshold
- For each threshold:
    - split data
    - compute weighted variance
- Choose best threshold

# Implementation Issues Encountered During Decision Tree Implementation

### 1. Iterate only to `len(data) - 1` when evaluating split points

When searching for the best split, the last element should not be considered as a split point because it would produce:

```text
Left : all samples
Right: empty
```

or vice versa.

Therefore, iterate only until:

```python
for i in range(len(data) - 1):
```

This guarantees there is at least one sample on both sides of the candidate split.

**Reason:**

* Prevents invalid splits.
* Avoids unnecessary computations.
* Prevents empty child nodes.

### 3. Pandas variance (`ddof`) vs NumPy variance

This is the point I would revise slightly.

Your note says:

> if we use pandas we must use ddof=0, better to use numpy

I would instead write:

> **For regression trees, use population variance (`ddof=0`). NumPy's `np.var()` already defaults to `ddof=0`, while Pandas' `Series.var()` defaults to `ddof=1`.**

Example:

```python
np.var(y)          # ddof=0
```

or

```python
df["score"].var(ddof=0)
```

**Reason:**

A tree node contains all samples assigned to that node, so we treat them as the population rather than a sample.

Also note:

* If you compute variance manually using loops,

```python
variance = sum((x-mean)**2)/len(x)
```

you are already computing **population variance**, so the `ddof` issue does not arise.

---

### 4. Always trap empty nodes before recursion

The first stopping condition should always be

```python
if df.empty:
    return None
```

or

```python
if len(data) == 0:
    return None
```

**Reason:**

Certain splits (especially simple median splits or invalid candidate splits) can create an empty child node.

Without this check, subsequent operations such as

```python
mean()
var()
```

or

```python
majority_class()
```

may return `NaN`, raise exceptions, or propagate invalid values into the recursive tree.

# CART Algorithm (Classification and Regression Tree)

## 1. Overview (intuition)

CART builds a decision tree by repeatedly asking:

> “What split best separates the data into purer groups?”

Think of it like repeatedly cutting a dataset with a knife so that each side becomes as “pure” as possible.

---

# 2. Core Idea

At each node:

* Try all possible feature splits
* Measure impurity reduction (e.g., Gini for classification, MSE for regression)
* Choose the best split
* Repeat recursively

---

# 3. CART Algorithm Steps (Classification)

## Step 1 — Start with full dataset

Given training data:

* Features: ( X )
* Labels: ( y )

This is the root node.

---

## Step 2 — For each feature, sort values

For each feature ( x_j ):

* Sort data by ( x_j )
* Consider only split points between **distinct values**

Example:

```
x: 1, 1, 2, 3, 5
```

Possible splits:

```
1.5, 2.5, 4.0
```

---

## Step 3 — Generate candidate splits

For each adjacent pair:

[
$$threshold = \frac{x_i + x_{i+1}}{2}$$
]

Only if:

[
x_i \ne x_{i+1}
]

---

## Step 4 — Split dataset

For each candidate threshold:

* Left node: ( x \le t )
* Right node: ( x > t )

---

## Step 5 — Compute impurity

For classification CART uses **Gini impurity**:

[
Gini = 1 - \sum p_k^2
]

For binary classification:

[
Gini = 1 - p_0^2 - p_1^2
]

---

## Step 6 — Compute weighted impurity

For each split:

[
G_{split} =
\frac{N_L}{N} G_L +
\frac{N_R}{N} G_R
]

Where:

* ( N_L, N_R ): number of samples in left/right node
* ( G_L, G_R ): Gini of each node

---

## Step 7 — Select best split

Choose split with:

[
\min G_{split}
]

Store:

* best feature
* best threshold
* best impurity

---

## Step 8 — Stop conditions

Stop splitting when:

* Node is pure (Gini = 0)
* Max depth reached
* Minimum samples per node reached
* No valid split improves impurity

---

## Step 9 — Split recursively

Repeat Steps 2–8 for:

* Left child node
* Right child node

---

## Step 10 — Final tree

The result is a binary tree:

* Internal nodes = decision rules
* Leaf nodes = class prediction (majority class)

---

# 4. CART Regression Variant

Same process, but replace Gini with:

## Mean Squared Error (MSE)

[
MSE = \frac{1}{n} \sum (y - \bar{y})^2
]

Split score:

[
MSE_{split} =
\frac{N_L}{N} MSE_L +
\frac{N_R}{N} MSE_R
]

---

# 5. Key Properties of CART

* Binary splits only
* Works for numerical and categorical (after encoding)
* Greedy (does NOT look ahead)
* Can overfit without pruning
* Handles non-linear decision boundaries

---

# 6. Pruning (important extension)

CART often uses **cost complexity pruning**:

[
R_\alpha(T) = R(T) + \alpha |T|
]

Where:

* ( R(T) ): training error
* ( |T| ): number of leaves
* ( \alpha ): penalty for complexity

---

# 7. Summary (short version)

At each node:

1. Sort data per feature
2. Try all valid split points
3. Compute impurity (Gini/MSE)
4. Choose best split
5. Recurse
6. Stop when stopping rules hit
